If a system is described by the discrete-time matrices:

$$A=\begin{bmatrix}0.1&0.2\\0.3&0.4\end{bmatrix}$$
$$B=\begin{bmatrix}0.5\\0.6\end{bmatrix}$$
$$C=\begin{bmatrix}0.7&0.8\end{bmatrix}$$
$$D=0.9$$

What's the observability matrix? What's the controllability matrix?

```octave
A = [0.1 0.2; 0.3 0.4]
B = [0.5;0.6]
C = [0.7,0.8]
D = 0.9
```

```octave
observability = [C; C*A]
```

```octave
controllability = [B A*B]
```

A discrete-time system is described by the (discrete-time) matrices

```octave
A = [0.1 0; 0 0.2]
B = [0.3;0.4]
C = [0.5 0]
D = [0.6]
```

```octave
[C; C*A]
```

The system is controllable but not observable
